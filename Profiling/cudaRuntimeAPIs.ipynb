```python
%%cuda

#include <iostream>
using namespace std;

int main()
{
    int nDevices;
    //return the number of compute capable devices
    cudaGetDeviceCount(&nDevices);
    for(int i = 0; i < nDevices; i++)
    {
        //structure that shows the device information
        cudaDeviceProp prop;
        //gives informaiton about each GPU device.
        cudaGetDeviceProperties(&prop, i);
        cout<<"Device number: "<<i<<endl;
        cout<<"Device name: "<<prop.name<<endl;
        cout<<"Number of SM's: "<<prop.multiProcessorCount<<endl;
        //add more to get more information. cudaDeviceProp has more information. Refer to cuda runtime API documentation. 

    }

}
```
